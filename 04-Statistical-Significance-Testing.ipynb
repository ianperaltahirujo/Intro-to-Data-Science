# Part 4

Simulation of 10,000 sequences of 200 coin **flips**

```python
num_simulations = 10000
sequence_length = 200

def count_changes():
    flips = np.random.choice(coin, sequence_length)
    changes = np.diff(flips == 'Heads')
    return np.count_nonzero(changes)
```

```python
change_counts = np.array([count_changes() for _ in range(num_simulations)])
change_stat = 100

plt.hist(change_counts, bins=np.arange(min(change_counts)-0.5, max(change_counts)+1.5, 1))
plt.scatter(change_stat, 0, color='red', s=50)

# Labels
plt.xlabel('Number of Changes')
plt.ylabel('Density')
plt.show()
```

# Part 5

Calculating p-value

```python
p_value = np.mean(change_counts <= change_stat)
p_value
```

# Part 6

Explanation:

The p-value of 0.55349 is larger than the traditional 0.05 threshold, meaning the observed change statistic is not unusual under the hypothesized model. Since a large p-value indicates that the observed data is consistent with randomness, we fail to reject the hypothesis that the flips were truly random and independent. The number of transitions (100) falls well within the expected range from simulations. Therefore, we do not have enough evidence to conclude that the sequence was generated by anything other than independent coin flips.
